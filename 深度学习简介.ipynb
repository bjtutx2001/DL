# 深度学习简介
深度学习（Deep Learning，DL）是机器学习的一个分支，主要通过多层人工神经网络，从数据中自动学习特征和规律，最终完成分类、回归、预测、生成等任务。
1. 特点
- 多层非线性变换
- 自动提取特特征
- 大数据和计算能力
- 可解释性差
2. 常见的深度学习模型
- 卷积神经网络（Convolutional Neural Networks）：擅长处理图片
- 循环神经网络（Recurrent）：擅长处理文本
- 自编码器（AutoEncoder）
- 生成对抗网络
- Transformer（自注意力架构）：擅长自然语言处理任务
- 深度强化学习
- 图神经网络（Graph）
- 可解释性神经网络
3. 深度学习生成架构
- PyTorch
- TensorFlow（过时）
4. PyTorch
- PyTorch是一个基于python的深度学习框架，它将数据封装成张量（Tensor）来进行处理
